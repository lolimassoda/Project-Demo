# Two-Way ANOVA: City and Program

Test the main effects of city and program type, plus their interaction.


```python
import numpy as np
import pandas as pd
import statsmodels.api as sm
from statsmodels.formula.api import ols
```

```python
rng = np.random.default_rng(42)
rows = []
for city, city_effect in {"Sudbury":-2,"London":2,"Toronto":5}.items():
    for program, program_effect in {"Standard":0,"Enhanced":6}.items():
        for _ in range(25):
            score = 70 + city_effect + program_effect
            if city == "Toronto" and program == "Enhanced":
                score += 2
            score += rng.normal(0, 5)
            rows.append([city, program, round(score, 2)])

df = pd.DataFrame(rows, columns=["city","program","score"])
df.head()
```

```python
model = ols("score ~ C(city) * C(program)", data=df).fit()
anova_table = sm.stats.anova_lm(model, typ=2)
anova_table
```

```python
anova_table.to_csv("anova_results.csv")
```
